**Chatalnawis**

```python
from sklearn.datasets import fetch_lfw_people

people = fetch_lfw_people(min_faces_per_person=5, resize=1)

n_samples, height, width = people.images.shape

print(n_samples, height, width)


X = people.data
n_features = X.shape[1]
'''
X.shape --> (1140, 2914) => (rijen=afbeeldingen, features = aantal pixels per afbeelding)
'''

#prediction op labels 
y = people.target #wat de algo moet kunnen voorspelen? (y bevat antwoorden) (elke rij in X hoort bij 1 element van y)
target_names = people.target_names
n_classes = target_names.shape[0] #tel hvl verschillende categorien zijn ze.

print(X)
print(n_features)

print(f"--------")
print(target_names)
print(y)
print(f"n_classes: {n_classes}")
```

```python
print(X.shape)

print(X.shape[0]) #aantal rijen of samples 
print(X.shape[1]) #aantal kolommen of features.
```

**Split data op in train en test**

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(X,y, train_size = 0.75, random_state = 42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Nu kan Dimension reductie toegepast worden

Bedoeling van dimReductionis om het aantal features en kolommen te verminderen. Om later enkel de belangrijke info bij te houden.

```python
X_train_scaled.shape
# """
# (4488, 11750)
# (n afbeeldingen, features/pixels per afbeelding)
# """
```

```python

#Hier bepaal je hoeveel nieuwe kenmerken PCA moet behouden.
#Het zijn 200 nieuwe combinaties van de oorspronkelijke pixelwaarden
n_components = 200 
```

***PCA aanmaken en trainen***

```python
from time import time

from sklearn.decomposition import PCA


t0 = time() #om te weten hvl tijd pca nodig heeft om te berekenen

pca = PCA(n_components,
          svd_solver='randomized',
          whiten=True).fit(X_train_scaled) #analyseer training data en leer de belangrijkste 200 componenten

# svd_solver='randomized'=> geeft een goede benadering vd belangrijkste datasets en is sneller
# whiten=True => zorgt ervoor dat de nieuwe componeten standardafwijking van 1 krijgen

print("Klaar in %0.3fs" % (time() - t0))
```

```python
X_train_scaled.shape
```

```python
eigenfaces = pca.components_.reshape((n_components, height, width))
eigenfaces[0]


# Je maakt dus van 200 lange rijen met getallen opnieuw 200 tweedimensionale afbeeldingen.

# pca.components_ = de geleerde patronen.
# eigenfaces = die patronen teruggezet naar afbeeldingsvorm.
# eigenfaces[0] = het eerste patroon als afbeelding.
```

**Nu gaat de 11750 om gezet worden naar 200**

```python
X_train_pca = pca.transform(X_train_scaled)
X_train_pca.shape
```

Fit() = pca leert de belangrijkste componete

transform() = pca zet de orginele om naar nieuw componeten

fit_transform() = doet beide achterelkaar
