```python
# nodige imports
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# data inladen
# in de datafile witten kolommen genaam Column1,2,3,4 die niet vermeld zijn en volledig leeg zijn dus die heb ik gedropt
# aangezien datum en tijdstip niet belangrijk zijn voor het bepalen van status_type drop ik die ook
df = pd.read_csv('./Live.xls')
df = df.drop(columns=['status_published','Column1','Column2','Column3','Column4'])
x = df.drop(columns=['status_type'])
y =  df['status_type']
```

```python
# dataset info
print(df.describe())

# verschillende classes 
class_0 = df[y == 'video']
class_1 = df[y == 'photo']
class_2 = df[y == 'link']
class_3 = df[y == 'status']
```

```python
# PCA
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
pca = PCA(n_components=0.95)
x_pca = pca.fit_transform(x_scaled)
original_dim = x.shape[1]
reduced_dim = x_pca.shape[1]
reduction_percent = (1 - reduced_dim / original_dim) * 100

print(f"Oorspronkelijke dimensies: {original_dim}")
print(f"Nieuwe dimensies (95% variantie): {reduced_dim}")
print(f"Datareductie: {reduction_percent:.2f}%")
```

```python
# kmeans
k = 4
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
y_pred = kmeans.fit_predict(x)
```

```python
# score
silhouette_score(x, kmeans.labels_)
```
