```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data_path = "data/secom/secom.data"
labels_path = "data/secom/secom_labels.data"

X = pd.read_csv(data_path, sep="\\s+", header=None)
y = pd.read_csv(labels_path, sep="\\s+", header=None, usecols=[0])

X = X.fillna(X.mean())

# data standardiseren
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.95)
X_pca_secom = pca.fit_transform(X_scaled)
n_components_secom = pca.n_components_
reduction_pct_secom = (1 - n_components_secom / X_scaled.shape[1])
```

```python
print("Componenten:", n_components_secom)
print("Vorm voor PCA:", X_scaled.shape)
print("Vorm na PCA:", X_pca_secom.shape)
print("Reductie:", round(reduction_pct_secom, 2), "%")
```
