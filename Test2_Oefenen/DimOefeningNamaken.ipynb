```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data_path = "data/secom/secom.data"
labels_path = "data/secom/secom_labels.data"

X = pd.read_csv(data_path, sep="\\s+", header=None)
y = pd.read_csv(labels_path, sep="\\s+", header=None, usecols=[0])

X = X.fillna(X.mean())

# data standardiseren
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.95)
X_pca_secom = pca.fit_transform(X_scaled)
n_components_secom = pca.n_components_
reduction_pct_secom = (1 - n_components_secom / X_scaled.shape[1])
```

```python
print("Componenten:", n_components_secom)
print("Vorm voor PCA:", X_scaled.shape)
print("Vorm na PCA:", X_pca_secom.shape)
print("Reductie:", round(reduction_pct_secom, 2), "%")
```

```python
pca_2 = PCA(n_components=2)
X_pca_2 = pca_2.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca_2[:, 0], X_pca_2[:, 1], c=y.values.ravel(), cmap="coolwarm", s=20)
plt.title("SECOM PCA (2D)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D 

pca_3 = PCA(n_components=3)
X_pca_3 = pca_3.fit_transform(X_scaled)

print(X_scaled.shape)
print(X_pca_3.shape)

fig = plt.figure(figsize=(9,6))
ax = fig.add_subplot(111, projection="3d")
ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

ax.scatter(X_pca_3[:, 0], X_pca_3[:, 1], X_pca_3[:, 2], c=y[0], s=20)

plt.show()


```

```python
print(pca_3.explained_variance_ratio_)
```
